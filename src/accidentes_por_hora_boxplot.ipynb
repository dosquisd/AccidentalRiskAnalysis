```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('dataset.csv')
df.head()
```

```python
df["HORA"]= pd.to_datetime(df["FECHA_OCURRENCIA"]).dt.hour
plt.figure(figsize=(10, 6))
sns.boxplot(x="CLASE",y="HORA", color="indianred", data = df)
plt.xticks(rotation=45, ha="right")
plt.yticks(ticks=range(0, 24, 2))
plt.grid(linestyle="-.")
```

```python
sns.boxplot(x="DIA_SEMANA_OCURRENCIA",y="HORA", palette="Set3", data = df)

plt.title("Distribución de horas de accidentes por día de la semana")
plt.xlabel("Dia de la semana")
plt.ylabel("Hora")
plt.xticks(rotation=45, ha="right")

plt.grid(linestyle="-.")
plt.show()
```
