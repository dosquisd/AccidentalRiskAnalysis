```python
import pandas as pd

PROCESSED_DATA: str = "../data/processed"
```

```python
df = pd.read_csv(f"{PROCESSED_DATA}/dataset.csv")
df["FECHA_OCURRENCIA"] = pd.to_datetime(df["FECHA_OCURRENCIA"])

# Agregando los días, meses y años por aparte
months = df["FECHA_OCURRENCIA"].dt.month
years = df["FECHA_OCURRENCIA"].dt.year
days = df["FECHA_OCURRENCIA"].dt.day

df["AÑO"] = years
df["DIA"] = days
df["MES"] = months

group_columns = [
    "AÑO",
    "DIA",
    "MES",
    "DIA_SEMANA_OCURRENCIA",
    "LOCALIDAD",
    "GRAVEDAD",
    "CLASE",
]

df
```

```python
count = df.groupby(group_columns).count()[["OBJECTID"]]
result_df = count.reset_index()
result_df = result_df.rename(columns={"OBJECTID": "CANTIDAD"})
result_df
```

```python
result_df.to_csv(f"{PROCESSED_DATA}/grouped.csv", index=False)
```
